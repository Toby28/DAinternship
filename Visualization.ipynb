```python
#Import the Library

import matplotlib.pyplot as plt
```

```python
#Basic Line Chart
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temp = [25, 26, 23, 25, 27, 24, 28]

plt.plot(days, temp)
plt.title('Daily Temperatures by Week')
plt.xlabel('Days')
plt.ylabel('Temperatures')
plt.show()
```

```python
#Customised Line Plot
plt.plot(days, temp, linestyle = '--', marker = 'p', color = 'blue')
plt.title('Daily Temperatures by Week')
plt.xlabel('Days')
plt.ylabel('Temperatures')
plt.show()
```

```python
# Sales Data Analysis - Bar Chart
products = ['Product A', 'Product B', 'Product C']
sales = [100, 150, 200]

plt.bar(products, sales, color = 'green', edgecolor = 'yellow')
plt.title('Sales Data Analysis')
plt.xlabel('Products')
plt.ylabel('Sales (in $)')
plt.show()
```

```python
# Market Share Analysis - Pie Chart
products = ['Product A', 'Product B', 'Product C']
share = [45, 30, 25]

plt.pie(share, labels=products, autopct = '%1.1f%%')
plt.title('Market Share Analysis')
plt.show()
```

```python
# Exam Score Analysis - Histogram
scores = [56, 47, 59, 66, 35, 37, 57, 66, 78, 79, 76, 72, 73, 63, 61, 40]

plt.hist(scores, bins = 5, color = 'yellow', edgecolor = 'red')
plt.title('Exam Score Analysis')
plt.xlabel('Score Analysis')
plt.ylabel('No of Students')
plt.show()
```

```python
# Height vs Weight Analysis - Scatter Plot
height = [150, 160, 165, 170, 175, 180, 185, 190]
weight = [50, 55, 56, 60, 65, 70, 72, 80]

plt.scatter(height, weight, color = 'pink')
plt.title('Height vs Weight Analysis')
plt.xlabel('Height (cm)')
plt.ylabel('Weight (kg)')
plt.show()
```

```python
#Saving a graph
plt.hist(scores, bins = 5, color = 'yellow', edgecolor = 'red')
plt.title('Exam Score Analysis')
plt.xlabel('Score Analysis')
plt.ylabel('No of Students')
plt.savefig('score.png')
```

```python
# Import Library
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load the dataset
tips = sns.load_dataset('tips')
tips.head()
```

```python
#Sactter Plot - Relation between totalbill and tip
sns.scatterplot(data = tips,  x='total_bill', y='tip', hue='time')
plt.title('Scatter PLot between Total Bill and Tips')
plt.show()
```

```python
#Bar Plot - Average bill by days
sns.barplot(data = tips,  x='day', y='total_bill', hue='time')
plt.title('Average bill by days')
plt.show()
```

We can say that on Thursday the total bill was high for Lunch and on Sunday the total bill was high for Dinner. Overall dinner shows higher demand than lunch

```python
#Count Plot - Frequency of Days
sns.countplot(data = tips,  x='day', palette='coolwarm')
plt.title('Frequency of days')
plt.show()
```

```python
#Box Plot - Total bill by days
sns.boxplot(data = tips,  x='day', y='total_bill', palette='Set2')
plt.title('Total bill by days')
plt.show()
```

```python
#Heat Map- Total bill by days
corr = tips.drop(columns=['sex', 'smoker', 'day', 'time']).corr()

sns.heatmap(corr,  annot = True, cmap='Set2')
plt.title('Heat Map')
plt.show()
```

```python
#Violin Plot - Total bill by days
sns.violinplot(data = tips,  x='day', y='total_bill', hue='sex', split=True, palette='muted')
plt.title('Violin Plot of total bills by gender')
plt.show()
```

```python
#Line Plot
import plotly.express as px

#Load Data
df = px.data.gapminder()
df.head()
```

```python
#Filter the data
df_us = df[df['country'] == 'United States']

#Create a Line Chart
fig = px.line(df_us, x='year', y='gdpPercap', title='GDP per Capita by Year')
fig.show()
```

```python
#Filter
countries = ['India', 'China', 'United States', 'Nigeria', 'Brazil']
df_count = df[df['country'].isin(countries)]

#Create a Line Chart
fig = px.line(df_count, x='year', y='pop', color='country',
              title='Population by Year', labels={'pop': 'Population', 'year': 'Year'})
fig.show()
```

```python
fig = px.scatter(df, x='gdpPercap', y='lifeExp', animation_frame ='year', animation_group='country',
                 size='pop', color='continent', hover_name='country',
                 title='GDP Per Capita vs Life Expectancy', log_x = True, size_max=60)

fig.show()
```

```python
fig = px.bar(df, x='country', y='pop', animation_frame ='year',
                 color='continent', hover_name='country',
                 title='Population by Country', range_y= [0, df['pop'].max()])

fig.show()
```

```python
fig = px.sunburst(tips, path=['day', 'time', 'sex'], values='total_bill',
                  title='Sunburst Chart')
fig.show()
```

```python
data = px.data.election()
geojson = px.data.election_geojson()

fig = px.choropleth(data, geojson=geojson, color='Bergeron', locations='district',
                    featureidkey='properties.district', projection='mercator',
                    color_continuous_scale='Viridis',
                    title='Election Result by District')
fig.show()
```

```python
fig = px.choropleth(df, color='gdpPercap', locations='iso_alpha',
                    hover_name='country',
                    color_continuous_scale=px.colors.sequential.Plasma,
                    title='Election Result by District')
fig.show()
```
