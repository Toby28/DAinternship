Dataset: Nigerian_Road_Traffic_Crashes_2020_2024.csv

About Data:
The dataset has several columns that describe various crash metrics per state for Q4 2020. These include:

- Quarter: Time period of the data collection.
- State: Nigerian states.
- Total_Crashes: Total number of crashes.
- Num_Injured: Number of injured individuals.
- Num_Killed: Number of fatalities.
- Total_Vehicles: Vehicles involved in crashes.
- Speed Violation (SPV): Speeding in road traffic crashes
- Driving Under Alcohol/Drug Influence (DAD):
- Poor Weather (PWR)
- Fatigue (FTQ): number of crashes attributed to driver fatigue
- Other_Factors


Objective: To provide a detailed analysis of traffic crashes in Nigeria, including the number of fatalities, injuries, and causes between 2020 and 2024.

Key Metrics: Total crashes, number of injured, number of killed, total vehicles involved, and contributing factors (speed violation, alcohol influence, fatigue, poor weather).

#Univariate Analysis


```python
# Histrogram of Distribution
plt.figure(figsize=(6,4))
sns.histplot(df['Total_Crashes'], kde=True, color='blue')
plt.title('Distribution of Total Crashes')
plt.show()
```

```python
# Boxplot for checking outliers
plt.figure(figsize=(6,4))
sns.boxplot(df['Total_Vehicles_Involved'], color='blue')
plt.title('Outliers distribution of Total vehicle Involved')
plt.show()
```

```python
#Count Plot for categorical column
plt.figure(figsize=(15,5))
sns.countplot(x='Quarter', data=df, hue='Quarter')
plt.title('Quarter frequency Distribution')
plt.show()
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(16, 5))
state_order = df['State'].value_counts().index
sns.countplot(x='State', data=df, order=state_order, palette='viridis')
for p in plt.gca().patches:
  plt.gca().annotate(f'{p.get_height()}', (p.get_x() + p.get_width() / 2., p.get_height()), ha='center',
                     va='baseline', fontsize=8, color='black', xytext=(0,5), textcoords= 'offset points')

plt.xticks(rotation=45, ha='right')
plt.title('Distribution of Crashes by States')
plt.tight_layout()
plt.show()
```

```python
#Count Plot for State
plt.figure(figsize=(15,5))
sns.barplot(x='State', y= 'Total_Crashes', data=df)
plt.title('State vs Total Crash')
plt.xticks(rotation=60, ha='right')
plt.show()
```

```python
#Let's split the quarter column and divide it into year and quarter
df['Year'] = df['Quarter'].apply(lambda x: x.split()[1])
df['Quarter'] = df['Quarter'].apply(lambda x: x.split()[0])
```

```python
#To create a violin plot that shows the distribution of crashes across states and quarters, you can use Seaborn’s violinplot.
# Grouping data by state to get the total crashes per state
top_10_states = df.groupby('State')['Total_Crashes'].sum().nlargest(10).index

# Filter the dataframe for only the top 10 states
df_top_10 = df[df['State'].isin(top_10_states)]

plt.figure(figsize=(15, 6))

# Line plot for Total Crashes across quarters for top 10 states
sns.lineplot(x='Quarter', y='Total_Crashes', hue='State', data=df_top_10, marker='o')

plt.title('Trend of Total Crashes for Top 10 States Across Quarters')
plt.xticks(rotation=60, ha='right')
plt.ylabel('Total Crashes')
plt.show()
```

- FCT (Federal Capital Territory) consistently has the highest number of crashes across the four quarters, though the crashes seem to decrease slightly over time.
- Kaduna, Nasarawa, and Niger are other states with relatively high crash counts, although they show more fluctuation from quarter to quarter.
- States like Lagos, Ogun, and Oyo have lower crash numbers in comparison, and they exhibit smaller variances, indicated by their narrower shaded areas.
- There is a general decline in crash numbers for most states as we move from Q4 to Q3, with some states like FCT and Kaduna maintaining their high crash rates.

```python
#Box Plot for Total Crashes according to Quarter
plt.figure(figsize=(15,5))
sns.boxplot(x='Quarter', y= 'Total_Crashes', data=df, hue='Quarter')
plt.title('Total Crashes according to Quarter')
plt.show()
```

- Each box shows the interquartile range (IQR), which represents the middle 50% of the data for each quarter (from the 25th percentile to the 75th percentile).
- The line inside each box represents the median crash count for that quarter, which is the middle value of the data.
- Width of the box: A narrower box indicates less variability in the crash numbers for that quarter, while a wider box shows more variability.

- Q4 (blue box) shows a higher median value for total crashes compared to the other quarters, indicating that more crashes tended to happen during this period.
- The outliers show that in all quarters, there are several states or regions where crashes are significantly higher than the typical range, with Q4 having the most extreme outlier over 500 crashes.
- The distribution of crashes for Q1, Q2, and Q3 is quite similar in terms of both the range and the median, but Q4 tends to have more variability and higher crash counts.

```python
#Scatter Plot for Total Crashes According to Year
plt.figure(figsize=(15,5))
sns.scatterplot(x='Total_Crashes', y= 'Num_Injured', data=df, hue='Year')
plt.title('Total Crashes According to Year')
plt.show()
```

- Data points are color-coded by year (2020 to 2024), and each year exhibits a similar trend of increasing injuries with more crashes.
- 2020 and 2021 have more scattered points, showing some cases with a higher number of injuries for fewer crashes compared to other years.
- 2022 and 2023 are more tightly clustered, reflecting a more predictable relationship between crashes and injuries.
- In 2024, there are fewer data points, which could suggest that data is still being collected or that the trend is incomplete.
- A few outliers are observed, especially in the years 2021 and 2023, where crashes above 400 led to a varying number of injuries. One particular outlier in 2021 shows a very high number of injuries (above 1000), which could indicate a significant crash incident.

**Insights from the Traffic Crash Analysis (2020-2024)**

- High Concentration of Crashes:
The majority of crashes occurred within the lower range (between 0 and 100 crashes per state), as shown by the distribution and boxplot analyses. However, certain states and periods, like Bauchi and FCT, experienced a significantly higher frequency of crashes.

- State-wise Trends:
A few states consistently show higher crash volumes, with Bauchi, FCT, Kaduna, Kogi, and Lagos leading the list. These states have maintained a higher crash count across multiple quarters, suggesting potential systemic issues or higher traffic densities.

- Quarterly Trends:
Across all states, there is no sharp increase or decrease in crashes quarter over quarter, indicating a relatively stable trend. However, some fluctuations do exist in states like FCT and Kaduna, suggesting seasonal or situational factors influencing these trends.

- Crashes vs. Injuries:
There is a strong positive correlation between total crashes and the number of injuries, which has remained consistent across all years (2020-2024). Larger crash incidents consistently result in higher numbers of injuries.

- Outliers and Exceptional Incidents:
Certain outliers in 2021 and 2023, particularly incidents with a high number of injuries (above 1000), indicate significant crash events that may have involved mass casualties or severe accidents. These should be investigated in further detail for causes and preventative measures.

**Recommendations:**

- Targeted Safety Measures for High-Risk States:
  - Bauchi, FCT, and Kaduna consistently show higher crash numbers. These states could benefit from:
  - Increased traffic surveillance, road safety audits, and stricter enforcement of traffic rules.

- Public awareness campaigns focused on reducing reckless driving, seatbelt usage, and accident reporting.

- Focus on Reducing Crash Severity:
  - The strong correlation between crashes and injuries suggests that many accidents involve personal injury.
  - Introducing stricter speed limits, implementing driver safety education, and enforcing vehicle safety inspections could help mitigate injury severity during crashes.

- Quarterly Traffic Data Analysis:
  - Continuous monitoring of quarterly crash data can help spot emerging trends. For instance, Q4 tends to see higher crash frequencies, which may be attributed to holiday traffic or weather conditions. Tailored interventions during these periods (e.g., road safety warnings, additional police presence) could reduce crash numbers.

- Deeper Investigation of Outliers:
  - Outliers such as crashes resulting in over 1000 injuries need to be analyzed further to determine the cause. Was it due to specific road conditions, weather, driver behavior, or vehicle failure? Understanding the root causes will help in designing future prevention strategies.

- Year-on-Year Crash Mitigation Strategies:
  - While the trend is consistent, exploring how other years' outliers (like in 2021 and 2023) affected overall statistics could offer insights into crash prevention. Benchmarking crash data and severity by year could reveal useful patterns to apply in future crash-prevention programs.
